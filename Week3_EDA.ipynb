```python
import pandas as pd

df = pd.read_csv("/content/ecommerce.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df['Satisfaction Level'].fillna(df['Satisfaction Level'].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

The Customer ID column was removed as it does
not contribute to behavioral analysis.

```python
df.drop('Customer ID', axis=1, inplace=True)
```

Duplicate records were checked and removed to ensure data integrity.

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

The dataset shows a relatively balanced distribution between male and female customers.

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x='Gender', data=df)
plt.title("Gender Distribution")
plt.show()
```

Premium members represent a significant portion of total customers.

```python
sns.countplot(x='Membership Type', data=df)
plt.title("Membership Type Distribution")
plt.xticks(rotation=45)
plt.show()
```

Most customers fall within the 25–40 age group.

```python
sns.histplot(df['Age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()
```

Spending behavior appears right-skewed, indicating a small group of high spenders.

```python
sns.histplot(df['Total Spend'], bins=20, kde=True)
plt.title("Total Spend Distribution")
plt.show()
```

There is a strong positive relationship between number of items purchased and total spending.

```python
sns.scatterplot(x='Items Purchased', y='Total Spend', data=df)
plt.title("Items Purchased vs Total Spend")
plt.show()
```

Customers who received discounts tend to show different spending behavior compared to those who did not.

```python
sns.boxplot(x='Discount Applied', y='Total Spend', data=df)
plt.title("Discount Impact on Spending")
plt.show()
```

Customers with lower satisfaction levels tend to have longer gaps between purchases.

```python
sns.boxplot(x='Satisfaction Level', y='Days Since Last Purchase', data=df)
plt.xticks(rotation=45)
plt.title("Satisfaction vs Purchase Recency")
plt.show()
```

Total Spend is strongly correlated with Items Purchased.
Days Since Last Purchase shows inverse relationship with spending.

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
